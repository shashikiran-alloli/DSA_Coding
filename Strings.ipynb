### Making translate table to create rules for character replacement and deletion (str.maketrans and s.translate) 

```python
# Example usage of str.maketrans and str.translate
# Create a translation table
translation_table = str.maketrans("abc", "123", "xyz")  # Replace 'a'->'1', 'b'->'2', 'c'->'3' and remove 'x', 'y', 'z'

# Input string
input_string = "abcxyzABCXYZ"

# Apply translation
translated_string = input_string.translate(translation_table)

# Output the result
print(f"Original String: {input_string}")
print(f"Translated String: {translated_string}")

# Using str.maketrans to create delete string punctuation - '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
s = '_Shashikiran@Channabasappa-}{}Alloli'
transtable = str.maketrans(string.punctuation, " "*len(string.punctuation), "")
print(f"Original String: {s}")
translated_string = s.translate(transtable)
print(f"Translated String: {translated_string}")
```

### Padding arbitrary characters at the left and right end of the string

```python
# Illustrating the usage of rjust and ljust in strings
# Input string
input_str = "Python"

# Padding the string with '*' to the left and right, making the total length len(input_str) + 5
padded_left = input_str.rjust(len(input_str) + 5, '*')
padded_right = input_str.ljust(len(input_str) + 5, '*')

# Output the results
print(f"Original String: {input_str}")
print(f"Left Padded String: {padded_left}")
print(f"Right Padded String: {padded_right}")
```

### 1. A string is subsequence of another string

```python
class Solution:
    def firstUniqChar(self, s: str) -> int:
        m = defaultdict(lambda: False)
        n = len(s)
        res = n
        for i in range(n-1, -1, -1):
            print(f"Character -> {s[i]}, bool -> {m[s[i]]}")
            if not m[s[i]]:
                res = i 
            m[s[i]] = True
        
        return -1 if res==n else res
        
```

## Longest Proper Prefix and Suffix (LPS) using KMP Algorithm
if str[i]==str[l] ; l=lps[i-1] then lps[i] = l+1

```python
class Solution:
    def longestPrefix(self, s: str) -> str:
        n = len(s)
        for m in range(n-1, -1, -1):
            for i in range(m):
                if s[i]!=s[n-m+i]:
                    break
            else:
                return s[:m]
        return ""

    def creatingLPS(self, s: str) -> str:
        lps = [0]
        l = 0
        n = len(s)
        i = 1
        while i<n:
            # print(f"i --> {i}")
            # print(f"\tLPS -> {lps}")
            # print(f"\tl->{l}, s[i]->{s[i]}, s[l]->{s[l]}")
            if s[i]==s[l]:
                l += 1
                lps.append(l)
                i += 1
            elif l==0:
                lps.append(0)
                i += 1
            else:
                # print(f"\tl=lps[l-1] -> {lps[l-1]}")
                l = lps[l-1]
                
        
        return lps
```

## Nth number made of prime digits

Difficulty: MediumAccuracy: 59.07%Submissions: 21K+Points: 4Average Time: 30m </br>
Given a number n. You have to find the nth number whose each digit is a prime number i.e. 2, 3, 5, 7. </br>
In other words, you have to find nth number of this sequence: 2, 3, 5, 7, 22, 23 ,... and so on.

Examples:

Input: n = 10 </br>
Output: 33 </br>
Explanation: 10th number in the sequence of numbers whose each digit is prime is 33. </br>
Input: n = 21 </br>
Output: 222 </br>
Explanation: The 21st number in the sequence of numbers whose each digit is prime is 222. </br>
Constraints: </br>
1 ≤ n ≤ 100 



```python
print(dir(deque))
```

```python
class Solution:
    def nthprimedigits_bfs(self, n):
        arr = [2, 3, 5, 7]
        count = 0
        q = deque()
        q.append(0)

        while q:
            f = q.popleft()
            for a in arr:
                t = f*10 + a
                q.append(t)
                count += 1
                # print(t, end=" ")
                if count == n:
                    return q.pop()

    def nthprimedigits_O_1(self, n):
        arr = ['0', '2', '3', '5', '7']
        l = 1
        prev_count = 0
        while True:
            l_count = 4**l
            curr_count = prev_count + l_count
            if prev_count < n and curr_count>=n:
                break
            l += 1
            prev_count = curr_count

        num = ""
        for i in range(1, l+1):
            for j in range(1, 5):
                if prev_count + 4**(l-i) < n:
                    prev_count += 4**(l-i)
                else:
                    print(arr[j], end="")
                    break

    def nthprimedigits_O_log4n(self, n):
        d = {0:'7', 1:'2', 2:'3', 3:'5'}
        num = ""
        while n>0:
            rem = n%4
            num += d[rem]
            n //= 4

        print(num[::-1])
                
```

```python
obj = Solution()
obj.nthprimedigits_bfs(100000)
```

## Permutation of a string

```python
s = 'abc'
```

```python
class Permutation():
    def permutation(self, s):
        res = []
        idx = 0
        n = len(s)
        def recursiveFunc(p, idx):
            if idx==n:
                res.append(''.join(p))
                return
            for i in range(idx, n):
                p[i], p[idx] = p[idx], p[i]
                recursiveFunc(p, idx+1)
                p[i], p[idx] = p[idx], p[i]

        recursiveFunc(list(s), 0)
        return res                
```

## Longest Duplicate Substring <a href="https://leetcode.com/problems/longest-duplicate-substring/description/" style="background-color: #00A300; color: #333; padding: 4px 8px; border-radius: 4px; text-decoration: none; font-weight: bold;">(LeetCode)</a>
Given a string s, consider all duplicated substrings: (contiguous) substrings of s that occur 2 or more times. The occurrences may overlap. </br>
Return any duplicated substring that has the longest possible length. If s does not have a duplicated substring, the answer is "".

Example 1:</br>
Input: s = "banana" </br>
Output: "ana"</br>

Example 2:</br>
Input: s = "abcd"</br>
Output: ""</br>
 

Constraints:</br>
2 <= s.length <= 3 * 104</br>
s consists of lowercase English letters.

```python
class LongestDuplicateSubstring:

    def longestPrefixSuffix(self, s: str) -> str:
        n = len(s)
        lps = [0]*n
        i = 1
        l = 0

        while i<n:
            if s[i]==s[l]:
                l += 1
                lps[i] = l
                i += 1
            elif l==0:
                lps[i] = l
                i += 1
            else:
                l = lps[l-1]
        
        res = p = 0
        for i in range(n):
            if res < lps[i]:
                res = lps[i]
                p = i
        
        return s[p-res+1:p+1]

    def longestDupSubstring_O_n2(self, s: str) -> str:
        maxlen = 0
        n = len(s)

        for i in range(n):
            temp = self.longestPrefixSuffix(s[i:])
            # print(f"For string -> {s[i:]}, the LPS -> {temp} with len -> {len(temp)}")
            if len(temp)>maxlen:
                maxlen = len(temp)
                res = temp
        
        return "" if maxlen==0 else res

    def longestDupSubstring_O_nlogn(self, s: str) -> str:
        n = len(s)
        a = 26
        mod = 10**9 + 7
        nums = [ord(k)-ord('a') for k in s]

        def search(L: int, a: int, mod: int, nums: List[int], n: int) -> int:
            h = 0 
            for k in range(L):
                h = (h*a + nums[k])%mod
            aL = pow(a, L, mod)                
            seen = defaultdict(list)
            seen[h].append(0)
            for start in range(1, n-L+1):
                h = (a*h - aL*nums[start-1] + nums[start+L-1])%mod
                if h in seen:
                    substring = nums[start:start+L]
                    if any(substring==nums[i:i+L] for i in seen[h]):
                        return start
                seen[h].append(start)
            return -1

        left = 1
        right = n-1
        start = -1
        while left <= right:
            L = (left+right)//2
            duplicate_start = search(L, a, mod, nums, n)
            if duplicate_start != -1:
                start = duplicate_start
                left = L+1
            else:
                right = L-1

        return s[start:start+left-1]
                    
```
