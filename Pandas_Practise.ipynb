```python
from functools import reduce, lru_cache
from collections import Counter, defaultdict, deque

import sys
import os
import numpy as np
import pandas as pd
```

```python
# Import findspark and initialize
import findspark
findspark.init()  # Auto-detects SPARK_HOME

# Import PySpark modules
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Create a SparkSession (your entry point to Spark)
spark = SparkSession.builder \
    .appName("JupyterPySparkApp") \
    .master("local[*]") \
    .config("spark.sql.repl.eagerEval.enabled", True) \
    .config("spark.sql.repl.eagerEval.maxNumRows", 20) \
    .getOrCreate()

# Verify
print(f"Spark Version: {spark.version}")
spark  # This shows the session info
```

```python
# Read a text file (Spark's README.md)
df = spark.read.text("/opt/spark/README.md")

# Basic operations
df_count = df.count()
print(f"Total lines: {df_count}")

# Filter and show
df_a = df.filter(col("value").contains("a"))
print(f"Lines with 'a': {df_a.count()}")

# Show sample (pretty-printed thanks to config)
df.show(5, truncate=False)
```
