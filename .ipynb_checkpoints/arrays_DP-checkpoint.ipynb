```python
class LongestCommonSubsequence:
    def __init__(self):
        self.res = 0

    
    def recursiveFunc(self, s1: str, s2: str) -> int:
        def func(s1,s2,n,m):
            if n==0 or m==0:
                return 0
            if s1[n-1]==s2[m-1]:
                return 1 + func(s1,s2,n-1,m-1)
            else:
                return max(func(s1,s2,n-1,m),func(s1,s2,n,m-1))

        return func(s1,s2,len(s1),len(s2))


    def memoizationFunc(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        memo = [[-1]*(m+1) for _ in range(n+1)]

        def func(s1,s2,n,m,memo):
            if memo[n][m]!=-1:
                return memo[n][m]
            
            if n==0 or m==0:
                memo[n][m] = 0
            elif s1[n-1] == s2[m-1]:
                memo[n][m] = 1 + func(s1,s2,n-1,m-1,memo)
            else:
                memo[n][m] = max(func(s1,s2,n-1,m,memo), func(s1,s2,n,m-1,memo))

            return memo[n][m]

        return func(s1,s2,n,m,memo)


    def memoizationFuncLRUCache(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)

        @functools.lru_cache(maxsize=None)
        def func(s1,s2,n,m):
            if n==0 or m==0:
                return 0
            elif s1[n-1] == s2[m-1]:
                return 1 + func(s1,s2,n-1,m-1)
            else:
                return max(func(s1,s2,n-1,m), func(s1,s2,n,m-1))

        return func(s1,s2,n,m)        


    def tabulationFunc(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        dp = [[0]*(m+1) for _ in range(n+1)]

        for i in range(1,n+1):
            for j in range(1,m+1):
                if s1[i-1] == s2[j-1]:
                    dp[i][j] = 1 + dp[i-1][j-1]
                else:
                    dp[i][j] = max(dp[i][j-1], dp[i-1][j])

        return dp[n][m]


    def spaceOptimizedDP(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        dp = [[0]*(m+1) for _ in range(2)]

        for i in range(1,n+1):
            for j in range(1,m+1):
                if s1[i-1] == s2[j-1]:
                    dp[i%2][j] = 1 + dp[(i-1)%2][j-1]
                else:
                    dp[i%2][j] = max(dp[i%2][j-1], dp[(i-1)%2][j])

        return dp[n%2][m]

    
    def spaceOptimizedDP_Ultimate(self, text1: str, text2: str) -> int:
        if len(text1) < len(text2):
            text1, text2 = text2, text1

        memo = [0] * len(text2)

        for ch in text1:
            count = 0
            for i in range(len(text2)):
                if count < memo[i]:
                    count = memo[i]
                elif ch == text2[i]:
                    memo[i] = count + 1

        return max(memo)


    def dynamicProgramming_1D(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        dp = [0] * (m + 1)

        for i in range(1, n+1):
            prev_diag = dp[0]
            for j in range(1, m+1):
                temp = dp[j]
                if s1[i-1]==s2[j-1]: 
                    dp[j] = prev_diag+1
                else:
                    dp[j] = max(dp[j], dp[j-1])
                prev_diag = temp

        return dp[m]
        
```

```python
#User function Template for python3
# https://www.geeksforgeeks.org/problems/perfect-sum-problem5633/1
class PerfectSumProblem:

    def recursiveFunc(self, arr, target):
        if target == 0:
            return 1
        if target < 0:
            return 0
        return 
        
    
    def perfectSum(self, arr, target):
        # code here
        dp = [0]*(target+1)
        dp[0]=1
        print(f"The DP array in the beginning -> {dp}")
        for n in arr:
            print(f"Current number being processed: {n}")
            for current_sum in range(target, n-1, -1):
                dp[current_sum] += dp[current_sum-n]
                print(f"Updated DP array after processing {n} for {current_sum} -> {dp}")

        return dp[target]

    # Below function returns values when you trying solve the problem with replacement (infinite availability of all elements in array) - Coin Change
    def CoinChange(self, arr, target):
        dp = [0]*(target+1)
        dp[0] = 1
        print(f"The DP array in the beginning -> {dp}")
        for n in arr:
            print(f"Current number being processed: {n}")
            for current_sum in range(n, target+1):
                dp[current_sum] += dp[current_sum - n]
                print(f"\tUpdated DP array after processing {n} for {current_sum} -> {dp}")

        return dp[target]
```

```python
# Coin change with minimum coins to sum 
# https://leetcode.com/problems/coin-change/description/
class CoinChange_MinCoins:
    def RecursiveFun(self, coins: List[int], amount: int) -> int:
       
        @lru_cache(maxsize=None)
        def recursive(amount, n):
            if amount == 0:
                return 0
            if n==0:
                return float('inf')
            if amount < 0:
                return float('inf')

            # return -1 if min(1 + recursiveFunc(amount-coins[n-1], n), recursiveFunc(amount, n-1)) == float('inf') else min(1 + recursiveFunc(amount-coins[n-1], n), recursiveFunc(amount, n-1))
            return min(1 + recursive(amount-coins[n-1], n), recursive(amount, n-1))
        
        res = recursive(amount, len(coins))   
        return -1 if res==float('inf') else res


    def dynamicProgramming2D(self, coins, amount):
        n = len(coins)

        dp = [[float('inf')]*(amount+1) for _ in range(n+1)]
        for i in range(n+1):
            dp[i][0] = 0

        for i in range(1, n+1):
            for j in range(1, amount+1):
                if coins[i-1] <= j:
                    dp[i][j] = min(dp[i][j-coins[i-1]] + 1, dp[i-1][j])
                else:
                    dp[i][j] = dp[i-1][j]
        
        return -1 if dp[n][amount]==float('inf') else dp[n][amount]


    def dynamicProgramming1D(self, coins: List[int], target: int) -> int:
        dp = [float('inf')]*(target+1)
        dp[0] = 0

        for val in coins:
            for i in range(val, target+1):
            # for i in range(target, val-1, -1):
                dp[i] = min(dp[i], dp[i-val]+1)

        return -1 if dp[target] == float('inf') else dp[target]

    
    def BitWise_DynamicProgramming(self, coins: List[int], target: int) -> int:
        count=0
        res=1<<target
        while res&1!=1:
                
            temp=0
            for i in coins:
                temp|=res>>i
            if temp==res:
                return -1
            res=temp
            count+=1
            
        return count
        
```

<h2>Edit Distance Problem <a href="https://leetcode.com/problems/edit-distance/" style="background-color: #00A300; color: #333; padding: 4px 8px; border-radius: 4px; text-decoration: none; font-weight: bold;">(LeetCode)</a></h2>
Given two strings word1 and word2, return the minimum number of operations required to convert word1 to word2.

You have the following three operations permitted on a word:

Insert a character</br>
Delete a character</br>
Replace a character
 

Example 1:</br>
Input: word1 = "horse", word2 = "ros"</br>
Output: 3</br>
Explanation: </br>
horse -> rorse (replace 'h' with 'r')</br>
rorse -> rose (remove 'r')</br>
rose -> ros (remove 'e')</br>

Example 2:</br>
Input: word1 = "intention", word2 = "execution"</br>
Output: 5</br>
Explanation: </br>
intention -> inention (remove 't')</br>
inention -> enention (replace 'i' with 'e')</br>
enention -> exention (replace 'n' with 'x')</br>
exention -> exection (replace 'n' with 'c')</br>
exection -> execution (insert 'u')

```python
class editDistance():
    
    def recursiveFunc(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        
        @lru_cache(maxsize=None)
        def func(s1, s2, n, m):
            if n==0:
                return m
            if m==0:
                return n
            if s1[n-1]==s2[m-1]:
                return func(s1, s2, n-1, m-1)
            return 1+min(func(s1, s2, n-1, m), func(s1, s2, n, m-1), func(s1, s2, n-1, m-1))

        return func(s1, s2, n, m)

    
    def dynamicProgramming_2D(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        dp = [[0]*(m+1) for _ in range(n+1)]

        for i in range(n+1):
            dp[i][0] = i
        for j in range(m+1):
            dp[0][j] = j
        for i in range(1, n+1):
            for j in range(1, m+1):
                if s1[i-1]==s2[j-1]:
                    dp[i][j] = dp[i-1][j-1]
                else:
                    dp[i][j] = 1 + min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])

        return dp[n][m]     


    def dynamicProgramming_1D(self, s1: str, s2: str) -> int:
        n, m = len(s1), len(s2)
        dp = [j for j in range(m+1)]
        for i in range(1, n+1):
            prev_diag = dp[0]
            dp[0] = i
            for j in range(1, m+1):
                temp = dp[j]
                if s1[i-1]==s2[j-1]:
                    dp[j] = prev_diag
                else:
                    dp[j] = 1 + min(prev_diag, dp[j-1], dp[j])
                prev_diag = temp

        return dp[m]                
```

## Longest Repeating Substring <a href="https://leetcode.com/problems/longest-repeating-substring/description/" style="background-color: #00A300; color: #333; padding: 4px 8px; border-radius: 4px; text-decoration: none; font-weight: bold;">(LeetCode)</a>

Given strings s1 and s2, return the minimum contiguous substring part of s1, so that s2 is a subsequence of the part. </br>
If there is no such window in s1 that covers all characters in s2, return the empty string "". If there are multiple such minimum-length windows, return the one with the left-most starting index.


Example 1:

Input: s = "abcd" </br>
Output: 0 </br>
Explanation: There is no repeating substring.</br>
Example 2:</br>

Input: s = "abbaba"</br>
Output: 2</br>
Explanation: The longest repeating substrings are "ab" and "ba", each of which occurs twice.</br>
Example 3:</br>

Input: s = "aabcaabdaab"</br>
Output: 3</br>
Explanation: The longest repeating substring is "aab", which occurs 3 times.</br>
 

Constraints:</br>

1 <= s.length <= 2000</br>
s consists of lowercase English letters.</br>

```python
class LongestRepeatingSubstring:
    def longestRepeatingSubstring_DP2D(self, s: str) -> int:
        n = len(s)
        res = 0
        dp = [[0]*(n+1) for _ in range(n+1)]

        for i in range(1, n+1):
            for j in range(i+1, n+1):
                if s[i-1] == s[j-1]: 
                    dp[i][j] = 1 + dp[i-1][j-1]
                    res = max(res, dp[i][j])
        
        return res


    def longestRepeatingSubstring_DP2D_Optimized(self, s: str) -> int:
        n = len(s)
        res = 0
        dp = [[0]*(n+1) for _ in range(2)]

        for i in range(1, n+1):
            for j in range(i+1, n+1):
                if s[i-1] == s[j-1]:
                    dp[i%2][j] = 1 + dp[(i-1)%2][j-1]
                    res = max(res, dp[i%2][j])
                else:
                    dp[i%2][j] = 0
        
        return res

    def longestRepeatingSubstring_DP1D(self, s: str) -> int:
        n = len(s)
        res = 0
        dp = [0]*(n+1)

        for i in range(1, n+1):
            prev_diag = dp[i]
            for j in range(i+1, n+1):
                temp = dp[j]
                if s[i-1] == s[j-1]:
                    dp[j] = 1 + prev_diag
                    res = max(res, dp[j])
                else:
                    dp[j] = 0
                prev_diag = temp

        return res
```

## Minimum Window Subsequence <a href="https://leetcode.com/problems/minimum-window-subsequence/description/" style="background-color: #00A300; color: #333; padding: 4px 8px; border-radius: 4px; text-decoration: none; font-weight: bold;">(LeetCode)</a>

Given strings s1 and s2, return the minimum contiguous substring part of s1, so that s2 is a subsequence of the part.</br>
If there is no such window in s1 that covers all characters in s2, return the empty string "". If there are multiple such minimum-length windows, return the one with the left-most starting index.</br>


Example 1:</br>
Input: s1 = "abcdebdde", s2 = "bde"</br>
Output: "bcde"</br>
Explanation: </br>
"bcde" is the answer because it occurs before "bdde" which has the same length.</br>
"deb" is not a smaller window because the elements of s2 in the window must occur in order.</br>

Example 2:</br>
Input: s1 = "jmeqksfrsdcmsiwvaovztaqenprpvnbstl", s2 = "u"</br>
Output: ""</br>
 
Constraints:</br>
1 <= s1.length <= 2 * 104</br>
1 <= s2.length <= 100</br>
s1 and s2 consist of lowercase English letters.

```python
class MinimumWindowSubsequence():
    def minimumWindowSubsequence_O_n2(self, s1: str, s2: str) -> str:
        n, m = len(s1), len(s2)
        dp = [[float('inf')]*(m+1) for _ in range(2)]
        for i in range(2):
            dp[i][0] = 0
        index = 0
        res = float('inf')
        for i in range(1, n+1):
            for j in range(1, m+1):
                if s1[i-1]==s2[j-1]:
                    dp[i%2][j] = 1 + dp[(i-1)%2][j-1]
                else:
                    dp[i%2][j] = 1 + dp[(i-1)%2][j]
            if dp[i%2][m]<res:
                res = dp[i%2][m]
                index = i

        return '' if res==float('inf') else s1[index-res:index]

    def minimumWindowSubsequence_O_n2_SpaceOptimized(self, s1: str, s2: str) -> str:
        n, m = len(s1), len(s2)
        dp = [float('inf')]*(m+1)
        dp[0] = 0
        prev_diag = dp[0]
        index = 0
        res = float('inf')
        for i in range(1, n+1):
            prev_diag = dp[0]
            for j in range(1, m+1):
                temp = dp[j]
                if s1[i-1]==s2[j-1]:
                    dp[j] = 1 + prev_diag
                else:
                    dp[j] = 1 + dp[j]
                prev_diag = temp
            if dp[m]<res:
                res = dp[m]
                index = i

        return '' if res==float('inf') else s1[index-res:index]
        
```
