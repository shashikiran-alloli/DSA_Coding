```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("CUDA version:", torch.version.cuda)
    print("GPU name:", torch.cuda.get_device_name(0))
    print("Number of GPUs:", torch.cuda.device_count())

    # Tensor Core support check (for Volta/Turing/Ampere+)
    major, minor = torch.cuda.get_device_capability(0)
    print("Compute Capability:", f"{major}.{minor}")
    if major >= 7:
        print("✅ Tensor Cores supported on this GPU")

    # Simple utilization test
    a = torch.randn(10000, 10000, device='cuda')
    b = torch.randn(10000, 10000, device='cuda')
    c = torch.matmul(a, b)
    torch.cuda.synchronize()
    print("✅ PyTorch successfully executed on GPU")
```

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
gpus = tf.config.list_physical_devices('GPU')
print("GPUs detected:", gpus)

if gpus:
    print("✅ TensorFlow GPU is available")
    print("GPU device details:")
    for gpu in gpus:
        details = tf.config.experimental.get_device_details(gpu)
        print(details)

    # Small TensorCore eligible op
    with tf.device('/GPU:0'):
        a = tf.random.normal([10000, 10000])
        b = tf.random.normal([10000, 10000])
        c = tf.matmul(a, b)
        print("✅ TensorFlow successfully executed on GPU")
else:
    print("⚠️ TensorFlow GPU NOT detected")
```
