## Rotten Oranges <a href="https://leetcode.com/problems/rotting-oranges/description/" style="background-color: #00A300; color: #333; padding: 4px 8px; border-radius: 4px; text-decoration: none; font-weight: bold;">(LeetCode)</a>

You are given an m x n grid where each cell can have one of three values:</br>

0 representing an empty cell,</br>
1 representing a fresh orange, or</br>
2 representing a rotten orange.</br>
Every minute, any fresh orange that is 4-directionally adjacent to a rotten orange becomes rotten.</br>

Return the minimum number of minutes that must elapse until no cell has a fresh orange. If this is impossible, return -1.</br>
 

Example 1:</br>
<img src="oranges.png" alt="Resized image" width="800" height="300"/>

Input: grid = [[2,1,1],[1,1,0],[0,1,1]]</br>
Output: 4</br>

Example 2:</br>
Input: grid = [[2,1,1],[0,1,1],[1,0,1]]</br>
Output: -1</br>
Explanation: The orange in the bottom left corner (row 2, column 0) is never rotten, because rotting only happens 4-directionally.</br>

Example 3:</br>
Input: grid = [[0,2]]</br>
Output: 0</br>
Explanation: Since there are already no fresh oranges at minute 0, the answer is just 0.


```python
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        fresh, min_count = 0, 0 
        n, m = len(grid), len(grid[0])
        rotten_oranges = deque()
        directions = [(0,1),(0,-1),(-1,0),(1,0)]

        for i in range(n):
            for j in range(m):
                if grid[i][j] == 1:
                    fresh += 1
                elif grid[i][j] == 2:
                    rotten_oranges.append((i,j))

        while rotten_oranges and fresh>0:
            l = len(rotten_oranges)
            for _ in range(l):
                r, c = rotten_oranges.popleft()
                for rx, cx in directions:
                    row, col = r+rx, c+cx
                    if row<0 or row==n or col<0 or col==m or grid[row][col]!=1:
                        continue
                    grid[row][col] = 2
                    rotten_oranges.append((row, col))
                    fresh -= 1
            min_count += 1

        return min_count if fresh==0 else -1
```

```python
obj = Solution()
```

```python
grid = [[2,1,1],[1,1,0],[0,1,1]]
obj.orangesRotting(grid)
```

```python
grid = [[2,1,1],[0,1,1],[1,0,1]]
obj.orangesRotting(grid)
```

```python
grid = [[0,2]]
obj.orangesRotting(grid)
```
